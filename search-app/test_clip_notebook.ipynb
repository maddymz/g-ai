## Image-to-Text Search
Search for text descriptions using an image query

```python
# Load embeddings to get image paths
data = load_embeddings()
print(f"Available images: {len(data['image_paths'])}")
for i, path in enumerate(data['image_paths'][:5]):
    print(f"  {i}: {path.split('/')[-1]}")
```

```python
# Search using the first image (apple)
query_image = data['image_paths'][0]
results = interactive_image_search(query_image, top_k=3)
```

```python
# Search using another image
query_image = data['image_paths'][3]  # vegetable basket
results = interactive_image_search(query_image, top_k=3)
```
